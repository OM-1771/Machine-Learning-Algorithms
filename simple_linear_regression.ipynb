```python
x = df['Weight'] # independent feature
y = df['Height'] # dependent feature
```

Train test Split of the Dataset 

```python
from sklearn.model_selection import train_test_split 
```

```python
x_train , x_test, y_train , y_test = train_test_split(x,y,test_size=0.20)
```

```python
x_train = x_train.values.reshape(-1, 1)
x_test  = x_test.values.reshape(-1, 1)
```

```python
x_train.shape  , x_test.shape , y_train.shape , y_test.shape
```

```python
#standarize the dataset train Train Independent data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
plt.scatter(x_test,x_test)
```

Train the Simple linear Regression Model

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train,y_train)

print("The slope or co-efficient of weight is : ",regressor.coef_)
print("Intercept: ",regressor.intercept_)
```

```python
plt.scatter(x_train,y_train)
```

```python
plt.plot(x_train,regressor.predict(x_train),'r')

#Now we got the best fit Line for our Model
```

-> Prediction of the Train Data 
-> Predicted height Output = intercep + coefficient (weight)



```python
y_pred_test = regressor.predict(x_test)
y_pred_test ,y_test
```

```python
plt.scatter(x_test , y_test)
```

```python
plt.scatter(x_test, regressor.predict(x_test), color='red')
```

Performanace Matrix
MSE , RMSE , MAE
R SQUARE AND ADJUSTED R SQUARE

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error , mean_squared_error
```

```python
mse = mean_squared_error(y_test,y_pred_test)
mae = mean_absolute_error(y_test,y_pred_test)
rmse = np.sqrt(mse)
```

```python
print("The Mean Sqaure Error is :",mse)
print("The Mean Absolute Error is : ",mae)
print("The Root Mean Square Error is : ",rmse)
```
